```python
import pyreadr
import pandas as pd
from sklearn.model_selection import train_test_split
```

```python
RANDOM_STATE = 42
TEST_SIZE = 0.3
```

# Einecke2010Kidney (Gene expression)
Source: https://github.com/statOmics/HDDAData/blob/master/data/Einecke2010Kidney.rda

```python
data = pyreadr.read_r("../Datasets/Einecke2010Kidney/Einecke2010Kidney.rda")["Einecke2010Kidney"]
print(data.shape)
display(data.head(3))
```

```python
data.rename(columns={"Reject_Status": "label"}, inplace=True)
data["label"] = data["label"].astype(int)
data.columns = [["label"] + list(range(len(data.columns))[:-1])]
data = data[list(range(len(data.columns))[:-1]) + ["label"]].copy()
data.columns = [col[0] if isinstance(col, tuple) else col for col in data.columns]
```

```python
train_data, test_data = train_test_split(data, test_size=TEST_SIZE, random_state=RANDOM_STATE, stratify=data["label"])
```

```python
train_data["subset"] = "train"
test_data["subset"] = "test"
data = pd.concat([train_data, test_data], axis=0).reset_index(drop=True)
data.columns = [col[0] if isinstance(col, tuple) else col for col in data.columns]
data["label"] = data["label"].astype(int)
data
```

```python
data.drop(columns=["subset"]).dtypes.unique(), data["label"].value_counts(normalize=True), data["subset"].value_counts(normalize=True), data.groupby("subset")["label"].value_counts(normalize=True), data.isna().sum().sum()
```

```python
data.to_parquet("../Benchmark/einecke2010kidney.parquet", index=False)
```

# Alon1999 (Gene expression)
Source: https://github.com/statOmics/HDDAData/blob/master/data/Alon1999.rda

```python
data = pyreadr.read_r("../Datasets/Alon1999/Alon1999.rda")["Alon1999"]
print(data.shape)
display(data.head(3))
```

```python
data.dtypes
```

```python
data.rename(columns={"Y": "label"}, inplace=True)
data.columns = [["label"] + list(range(len(data.columns))[:-1])]
data = data[list(range(len(data.columns))[:-1]) + ["label"]].copy()
data.columns = [col[0] if isinstance(col, tuple) else col for col in data.columns]
```

```python
data.loc[data["label"] == "n", "label"] = 0
data.loc[data["label"] == "t", "label"] = 1
data["label"] = data["label"].astype(int)
data.head(3)
```

```python
train_data, test_data = train_test_split(data, test_size=TEST_SIZE, random_state=RANDOM_STATE, stratify=data["label"])
```

```python
train_data["subset"] = "train"
test_data["subset"] = "test"
data = pd.concat([train_data, test_data], axis=0).reset_index(drop=True)
data.columns = [col[0] if isinstance(col, tuple) else col for col in data.columns]
data["label"] = data["label"].astype(int)
data
```

```python
data.drop(columns=["subset"]).dtypes.unique(), data["label"].value_counts(normalize=True), data["subset"].value_counts(normalize=True), data.groupby("subset")["label"].value_counts(normalize=True), data.isna().sum().sum()
```

```python
data.to_parquet("../Benchmark/alon1999.parquet", index=False)
```
